```python
for column in ["Milieu", "Région", "Sexe"]:

    print(f"\n{column} categories:")
    print(df[column].value_counts(dropna=False))
```

```python
year_cols = [
    column for column in df.columns
    if str(column).isdigit()
]

for column in year_cols:

    numeric_values = pd.to_numeric(
        df[column],
        errors="coerce"
    )

    invalid_values = df.loc[
        df[column].notna() & numeric_values.isna(),
        column
    ].unique()

    print(f"\nYear: {column}")
    print("Original data type:", df[column].dtype)
    print("Invalid values:", invalid_values)
    print("Invalid count:", len(invalid_values))
```

```python
values_2024 = pd.to_numeric(
    df["2024"],
    errors="coerce"
)

zero_rows = df.loc[
    values_2024.eq(0),
    ["Milieu", "Région", "Sexe", "2024"]
]

print("Number of zero values:", len(zero_rows))

display(zero_rows)
```
