## 3. Creating Our First Cleaned Dataset

```python
from pathlib import Path
import pandas as pd
import numpy as np

# 1. Set project paths
PROJECT_ROOT = Path.cwd()

if PROJECT_ROOT.name == "notebooks":
    PROJECT_ROOT = PROJECT_ROOT.parent

RAW_DIR = PROJECT_ROOT / "data" / "raw"
PROCESSED_DIR = PROJECT_ROOT / "data" / "processed"

DATA_FILE = RAW_DIR / "i_i3287_data.xlsx"

# 2. Reload the original dataset
df_raw = pd.read_excel(
    DATA_FILE,
    sheet_name="Data",
    header=1
)

df_raw.columns = df_raw.columns.astype(str).str.strip()

# 3. Create a separate working copy
df_clean = df_raw.copy()

# 4. Identify columns
id_columns = ["Milieu", "Région", "Sexe"]

year_cols = [
    column for column in df_clean.columns
    if str(column).isdigit()
]

# 5. Clean categorical values
for column in id_columns:
    df_clean[column] = (
        df_clean[column]
        .astype("string")
        .str.strip()
    )

# 6. Replace dash markers with missing values
# Then convert yearly columns to numeric
for column in year_cols:
    df_clean[column] = df_clean[column].replace("-", np.nan)

    df_clean[column] = pd.to_numeric(
        df_clean[column],
        errors="coerce"
    )

print("Basic cleaning completed!")
print("Dataset shape:", df_clean.shape)
print("Year columns converted to numeric.")
```

## 3-2. Apply the six documented 2024 corrections

This code checks that each original value really is zero before replacing it. It will stop rather than silently overwrite an unexpected value

```python
# Official HCP values for 2024
# Source: HCP, Tableau 24, unemployment by age, sex and residence

correction_values = {
    ("National", "Masculin"): 11.6,
    ("National", "Féminin"): 19.4,
    ("Urbain", "Masculin"): 14.3,
    ("Urbain", "Féminin"): 25.3,
    ("Rural", "Masculin"): 6.7,
    ("Rural", "Féminin"): 6.8
}

corrections = []

for (milieu, sexe), new_value in correction_values.items():

    mask = (
        (df_clean["Milieu"] == milieu)
        & (df_clean["Région"] == "Ensemble")
        & (df_clean["Sexe"] == sexe)
    )

    if mask.sum() != 1:
        raise ValueError(
            f"Expected exactly one row for {milieu}, {sexe}"
        )

    old_value = df_clean.loc[mask, "2024"].iloc[0]

    if pd.isna(old_value) or old_value != 0:
        raise ValueError(
            f"Unexpected original value for {milieu}, {sexe}: {old_value}"
        )

    # Apply the documented correction
    df_clean.loc[mask, "2024"] = new_value

    # Record the correction
    corrections.append({
        "Milieu": milieu,
        "Région": "Ensemble",
        "Sexe": sexe,
        "Year": 2024,
        "Original_value": old_value,
        "Corrected_value": new_value,
        "Reason": "Verified against official HCP publication"
    })

corrections_log = pd.DataFrame(corrections)

print("Number of documented corrections:", len(corrections_log))
display(corrections_log)
```

```python
# Check data types
print("Year column data types:")
print(df_clean[year_cols].dtypes)

# Count missing values by year
quality_summary = pd.DataFrame({
    "Available values": df_clean[year_cols].notna().sum(),
    "Missing values": df_clean[year_cols].isna().sum()
})

print("\nData quality summary:")
display(quality_summary)

# Check whether categorical combinations remain unique
print(
    "\nDuplicate combinations:",
    df_clean.duplicated(
        subset=id_columns
    ).sum()
)

print(
    "\nTotal missing values:",
    df_clean[year_cols].isna().sum().sum()
)
```

 # 4. Reshape the dataset

 So For analysis and visualization, we'll convert it to long format: one row per year and category combination.

```python
# Convert wide format into long format

df_long = df_clean.melt(
    id_vars=["Milieu", "Région", "Sexe"],
    value_vars=year_cols,
    var_name="Année",
    value_name="Taux_chomage"
)

# Convert year values into integers
df_long["Année"] = df_long["Année"].astype(int)

# Sort the data
df_long = df_long.sort_values(
    by=["Milieu", "Région", "Sexe", "Année"]
).reset_index(drop=True)

print("Reshaping completed!")
display(df_long.head(10))
```

```python
print("Dataset shape:", df_long.shape)

print("\nColumn names:")
print(df_long.columns.tolist())

print("\nData types:")
print(df_long.dtypes)

print("\nMissing unemployment values:")
print(df_long["Taux_chomage"].isna().sum())

print("\nDuplicate records:")
print(
    df_long.duplicated(
        subset=["Milieu", "Région", "Sexe", "Année"]
    ).sum()
)
```

# 5. First Data Analysis

```python
# Filter national unemployment rates
national_total = df_long[
    (df_long["Milieu"] == "National") &
    (df_long["Région"] == "Ensemble") &
    (df_long["Sexe"] == "Total")
].copy()

# Sort by year
national_total = national_total.sort_values("Année")

display(national_total[["Année", "Taux_chomage"]])
```

# 5-2 visualize the evolution
first exploratory data visualization.

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(10, 5))

plt.plot(
    national_total["Année"],
    national_total["Taux_chomage"],
    marker="o"
)

plt.title("Evolution du taux de chômage national au Maroc (2015-2025)")
plt.xlabel("Année")
plt.ylabel("Taux de chômage (%)")

plt.xticks(national_total["Année"], rotation=45)
plt.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
```

```python
# Select regional total unemployment rates for 2025
regional_2025 = df_long[
    (df_long["Milieu"] == "National") &
    (df_long["Région"] != "Ensemble") &
    (df_long["Sexe"] == "Total") &
    (df_long["Année"] == 2025)
][["Région", "Taux_chomage"]].copy()

# Sort regions by unemployment rate
regional_2025 = regional_2025.dropna().sort_values(
    "Taux_chomage",
    ascending=True
)

display(regional_2025)
```

# 5-3 Compare unemployment across regions in 2025


```python
plt.figure(figsize=(10, 6))

plt.barh(
    regional_2025["Région"],
    regional_2025["Taux_chomage"]
)

plt.title("Taux de chômage par région au Maroc (2025)")
plt.xlabel("Taux de chômage (%)")
plt.ylabel("Région")

plt.grid(axis="x", alpha=0.3)
plt.tight_layout()
plt.show()
```
